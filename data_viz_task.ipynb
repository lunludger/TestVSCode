# Data visualisation task 

**Additional information**  
[Cars93 metadata](https://jse.amstat.org/v1n1/datasets.lock.html) 

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
df = pd.read_csv('Cars93.csv', index_col=0, keep_default_na=False)

```

### 1. Boxplot

```python
car_types = ['Audi', "Hyundai", "Suzuki", "Toyota"]
filter_df = df[df['Manufacturer'].isin(car_types)]

# Create the box plot
plt.figure()
sns.boxplot(
   x ="Manufacturer",
   y ="RPM",
   data = filter_df
)

# Title and labels
plt.xlabel("Manufacturer")
plt.ylabel("RPM")
plt.title("Boxplot showing Manufacturer and RPM")

# Show plot
plt.show()
```

**Boxplot question:**  
Which of these manufacturers has the car with the highest revs per mile?


**Boxplot answer:**
Hyundai has the highest Revs per mile

### 2. Histogram

```python
# Histogram of MPG
sns.histplot(df["MPG.city"], label="City MPG")
sns.histplot(df['MPG.highway'], label="Highway MPG")

# Title and labels
plt.xlabel("MPG")
plt.ylabel("Efficiency")
plt.title("Histogram of Highway vs City")

# Legend
plt.legend()

# Show plot
plt.show()
```

**Histogram question:**  
Is it generally more fuel efficient to drive in the city or on the highway?

**Histogram answer:**
It is more fuel efficient to drive in the highway

### 3. Lineplot

```python
# Lineplot (Wheelbase vs turning circle)
# Sort by Wheelbase for a proper line plot
df_sorted = df.sort_values(by="Wheelbase")

# Create line plot
plt.figure()
plt.plot(df_sorted["Wheelbase"], df_sorted["Turn.circle"], marker='o')

# Labels and title
plt.xlabel("Wheelbase")
plt.ylabel("Turning Circle")
plt.title("Turning circle vs Wheelbase")

# Show plot
plt.show()
```

**Lineplot question:**  
What is this relationship? What happens when the wheelbase gets larger?


**Lineplot answer:**  
The larger the wheelbase the wider the turning circle

### 4. Barplot

```python
# Barplot for mean horsepower
# Group by type and calculate mean horsepower
mean_horsepower = df.groupby("Type")["Horsepower"].mean()

# Create bar plot
plt.figure()
mean_horsepower.plot(kind="bar")

# Labels and title
plt.xlabel("car type")
plt.ylabel("Mean power")
plt.title("Car type vs Horsepower")

# Show plot
plt.show()
```

**Barplot question:**  
Does a larger car mean more horsepower?



**Barplot answer:**
A larger car mean more horsepower

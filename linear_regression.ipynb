The Diabetes dataset from scikit-learn includes these two features amongst others
s1: total serum cholestorol
s2: low-density lipoproteins (a type of cholesterol)
By the nature of these features, there is natutally a correlation between them. This is a nice practice for linear regression
This data can simply be loaded as a pandas DataFrame in the following way

```python
# Import libraries
from sklearn.datasets import load_diabetes
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load dataset
df = load_diabetes(as_frame=True).data[['s1', 's2']]

# Plot data
plt.figure()
sns.scatterplot(data=df, x='s1', y='s2')
plt.show()
plt.close()
```

To create our model, we import LinearRegression from the linear_model module in sklearn:

To allow compatibility, we mus ensure that our model inputs are in the shape (n_samples, n_features)
Because there is one feature, it will be (n_samples, 1)

```python
# Reshape the 's1' column into a 2D array for model input
X = df['s1'].values.reshape(-1, 1)

# Extract the 's2' column as the target variable
y = df['s2'].values
```

Now that we have our data in the desired shape, we can fit the data and make predictions

```python
# Fit the linear regression model to the data
simple_model = LinearRegression()
simple_model.fit(X, y)

# Predict target values based on input data
y_pred = simple_model.predict(X)
```

Lets take a look at our model performance using a plot

```python
# Create a new figure for plotting
plt.Figure()

# Plot the scatter plot of the original data
sns.scatterplot(data=df, x='s1', y='s2')

# Plot the linear regression line based on the prediction
plt.plot(X, y_pred)

# Display and close the plot
plt.show()
plt.close()
```
